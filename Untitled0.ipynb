```
from IPython.display import HTML

html_content = """
<!DOCTYPE html>
<html lang="pt-BR">
<head>
<meta charset="UTF-8">
<title>Snake.io Clone</title>
<style>
  body {
    margin: 0;
    overflow: hidden;
    background-color: #111;
    font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
    user-select: none;
  }
  #game-container {
    position: relative;
    width: 100vw;
    height: 100vh;
  }
  canvas {
    display: block;
    background-color: #1a1a1a;
  }
  #ui {
    position: absolute;
    top: 10px;
    left: 10px;
    color: #fff;
    text-shadow: 2px 2px 2px #000;
    pointer-events: none;
  }
  #score-board {
    font-size: 24px;
    font-weight: bold;
  }
  #kill-board {
    font-size: 20px;
    color: #ff3366;
    font-weight: bold;
    margin-top: 5px;
  }
  #leaderboard {
    position: absolute;
    top: 10px;
    right: 10px;
    background: rgba(0, 0, 0, 0.6);
    color: #fff;
    padding: 15px;
    border-radius: 8px;
    font-size: 14px;
    min-width: 180px;
    pointer-events: none;
    border: 1px solid rgba(255, 255, 255, 0.2);
  }
  #leaderboard h3 {
    margin: 0 0 10px 0;
    text-align: center;
    font-size: 16px;
    color: #00ffcc;
    text-transform: uppercase;
  }
  #leaderboard ol {
    margin: 0;
    padding-left: 20px;
  }
  #leaderboard li {
    margin-bottom: 5px;
  }
  #game-over-screen {
    position: absolute;
    top: 50%;
    left: 50%;
    transform: translate(-50%, -50%);
    background: rgba(0, 0, 0, 0.85);
    color: #fff;
    padding: 30px 50px;
    border-radius: 15px;
    text-align: center;
    border: 2px solid #00ffcc;
    display: none;
  }
  #game-over-screen h1 {
    margin-top: 0;
    color: #ff3366;
  }
  #play-btn {
    background: #00ffcc;
    color: #000;
    border: none;
    padding: 10px 25px;
    font-size: 18px;
    font-weight: bold;
    border-radius: 5px;
    cursor: pointer;
    margin-top: 15px;
    transition: 0.2s;
  }
  #play-btn:hover {
    background: #00b38f;
    transform: scale(1.05);
  }
</style>
</head>
<body>

<div id="game-container">
  <div id="ui">
    <div id="score-board">Comprimento: <span id="score-val">0</span></div>
    <div id="kill-board">Kills: <span id="kill-val">0</span></div>
    <div style="font-size: 13px; color: #aaa; margin-top: 5px;">Controles: Use <b>WASD</b> ou o Mouse para guiar!</div>
    <div style="font-size: 13px; color: #aaa; margin-top: 3px;">Segure o <b>Clique Esquerdo</b> ou o <b>Espaço</b> para Acelerar!</div>
  </div>

  <div id="leaderboard">
    <h3>Classificação</h3>
    <ol id="leaderboard-list"></ol>
  </div>

  <div id="game-over-screen">
    <h1>Você Morreu!</h1>
    <p>Sua pontuação final foi: <span id="final-score">0</span></p>
    <p>Eliminações: <span id="final-kills">0</span></p>
    <button id="play-btn">Jogar Novamente</button>
  </div>

  <canvas id="gameCanvas"></canvas>
</div>

<script>
  const canvas = document.getElementById('gameCanvas');
  const ctx = canvas.getContext('2d');
  const scoreVal = document.getElementById('score-val');
  const killVal = document.getElementById('kill-val');
  const leaderboardList = document.getElementById('leaderboard-list');
  const gameOverScreen = document.getElementById('game-over-screen');
  const finalScore = document.getElementById('final-score');
  const finalKills = document.getElementById('final-kills');
  const playBtn = document.getElementById('play-btn');

  // Ajustar tamanho do canvas
  function resizeCanvas() {
    canvas.width = window.innerWidth;
    canvas.height = window.innerHeight;
  }
  window.addEventListener('resize', resizeCanvas);
  resizeCanvas();

  // Configurações do Mundo
  const WORLD_WIDTH = 3000;
  const WORLD_HEIGHT = 3000;
  const NUM_BOTS = 15;
  const NUM_FOOD = 400;

  let camera = { x: 0, y: 0 };
  let mouse = { x: 0, y: 0 };
  let isBoosting = false;
  let player;
  let snakes = [];
  let foods = [];
  let isGameOver = false;
  let playerKills = 0;
  let keysPressed = {};

  // Cores divertidas para as cobras
  const colors = [
    '#00ffcc', '#ff3366', '#ffcc00', '#33ccff', '#ff66cc',
    '#99ff33', '#ff9933', '#cc66ff', '#00ff66', '#ffff00'
  ];

  class Snake {
    constructor(x, y, color, name, isBot = true) {
      this.name = name;
      this.color = color;
      this.isBot = isBot;
      this.segments = [];
      this.angle = Math.random() * Math.PI * 2;
      this.baseSpeed = 3.5;
      this.speed = this.baseSpeed;
      this.radius = 12;
      this.length = 30; // comprimento inicial desejado

      // Inicializa os segmentos em uma fila
      for (let i = 0; i < 20; i++) {
        this.segments.push({ x: x - i * 5, y: y });
      }
    }

    update() {
      let head = this.segments[0];

      if (this.isBot) {
        // IA SUPER AGRESSIVA
        let prey = null;
        let minPreyDst = 250; // Raio de detecção para atacar
        let shouldBoost = false;

        // 1. Procurar o jogador ou outros bots para atacar e cortar a frente
        for (let other of snakes) {
          if (other === this) continue;
          let otherHead = other.segments[0];
          let dst = Math.hypot(otherHead.x - head.x, otherHead.y - head.y);
          if (dst < minPreyDst) {
            minPreyDst = dst;
            prey = other;
          }
        }

        if (prey) {
          let preyHead = prey.segments[0];
          let targetX = preyHead.x + Math.cos(prey.angle) * (prey.speed * 12);
          let targetY = preyHead.y + Math.sin(prey.angle) * (prey.speed * 12);

          this.angle = Math.atan2(targetY - head.y, targetX - head.x);

          if (minPreyDst < 150 && this.length > 25) {
            shouldBoost = true;
          }
        } else {
          let foodTarget = this.getNearestFood(head);
          if (foodTarget) {
            this.angle = Math.atan2(foodTarget.y - head.y, foodTarget.x - head.x);
          } else if (Math.random() < 0.05) {
            this.angle += (Math.random() - 0.5) * 1.5;
          }
        }

        if (shouldBoost) {
          this.speed = this.baseSpeed * 1.8;
          this.length -= 0.06;
          if (Math.random() < 0.15) {
            let tail = this.segments[this.segments.length - 1];
            foods.push(new Food(tail.x, tail.y, 3, this.color));
          }
        } else {
          this.speed = this.baseSpeed;
        }

      } else {
        // Movimentação híbrida: Teclado (WASD) ou Mouse
        let keyboardX = 0;
        let keyboardY = 0;

        if (keysPressed['KeyW'] || keysPressed['Keyw'] || keysPressed['ArrowUp']) keyboardY -= 1;
        if (keysPressed['KeyS'] || keysPressed['Keys'] || keysPressed['ArrowDown']) keyboardY += 1;
        if (keysPressed['KeyA'] || keysPressed['Keya'] || keysPressed['ArrowLeft']) keyboardX -= 1;
        if (keysPressed['KeyD'] || keysPressed['Keyd'] || keysPressed['ArrowRight']) keyboardX += 1;

        if (keyboardX !== 0 || keyboardY !== 0) {
          this.angle = Math.atan2(keyboardY, keyboardX);
        } else {
          // Caso não use teclado, segue o Mouse
          let targetX = head.x - camera.x + (mouse.x - canvas.width / 2);
          let targetY = head.y - camera.y + (mouse.y - canvas.height / 2);
          this.angle = Math.atan2(targetY - head.y, targetX - head.x);
        }

        // Jogador acelera pressionando o clique esquerdo ou a tecla de Espaço
        let activeBoost = isBoosting || keysPressed['Space'] || keysPressed[' '];
        if (activeBoost && this.length > 20) {
          this.speed = this.baseSpeed * 1.8;
          this.length -= 0.08;
          if (Math.random() < 0.15) {
            let tail = this.segments[this.segments.length - 1];
            foods.push(new Food(tail.x + (Math.random() - 0.5) * 10, tail.y + (Math.random() - 0.5) * 10, 3, this.color));
          }
        } else {
          this.speed = this.baseSpeed;
        }
      }

      // Calcular nova cabeça
      let newHead = {
        x: head.x + Math.cos(this.angle) * this.speed,
        y: head.y + Math.sin(this.angle) * this.speed
      };

      // Limites do mapa
      if (newHead.x < 0) newHead.x = 0;
      if (newHead.x > WORLD_WIDTH) newHead.x = WORLD_WIDTH;
      if (newHead.y < 0) newHead.y = 0;
      if (newHead.y > WORLD_HEIGHT) newHead.y = WORLD_HEIGHT;

      this.segments.unshift(newHead);

      let maxSegments = Math.floor(this.length);
      while (this.segments.length > maxSegments) {
        this.segments.pop();
      }
    }

    getNearestFood(head) {
      let minDst = Infinity;
      let nearest = null;
      for (let f of foods) {
        let dst = Math.hypot(f.x - head.x, f.y - head.y);
        if (dst < minDst && dst < 400) {
          minDst = dst;
          nearest = f;
        }
      }
      return nearest;
    }

    draw() {
      ctx.save();
      for (let i = this.segments.length - 1; i >= 0; i--) {
        let seg = this.segments[i];
        let scaleRadius = this.radius * (1 - (i / this.segments.length) * 0.3);

        ctx.beginPath();
        ctx.arc(seg.x - camera.x, seg.y - camera.y, scaleRadius, 0, Math.PI * 2);
        ctx.fillStyle = this.color;
        ctx.shadowBlur = i === 0 ? 10 : 0;
        ctx.shadowColor = this.color;
        ctx.fill();
        ctx.closePath();

        if (i === 0) {
          ctx.shadowBlur = 0;
          ctx.fillStyle = '#fff';
          let eyeOffset = 6;

          ctx.beginPath();
          ctx.arc(
            seg.x - camera.x + Math.cos(this.angle - 0.5) * eyeOffset,
            seg.y - camera.y + Math.sin(this.angle - 0.5) * eyeOffset,
            4, 0, Math.PI * 2
          );
          ctx.fill();
          ctx.closePath();

          ctx.beginPath();
          ctx.arc(
            seg.x - camera.x + Math.cos(this.angle + 0.5) * eyeOffset,
            seg.y - camera.y + Math.sin(this.angle + 0.5) * eyeOffset,
            4, 0, Math.PI * 2
          );
          ctx.fill();
          ctx.closePath();
        }
      }

      let head = this.segments[0];
      ctx.font = 'bold 12px Arial';
      ctx.fillStyle = '#fff';
      ctx.textAlign = 'center';
      ctx.fillText(this.name, head.x - camera.x, head.y - camera.y - 20);
      ctx.restore();
    }
  }

  class Food {
    constructor(x, y, size, color) {
      this.x = x || Math.random() * WORLD_WIDTH;
      this.y = y || Math.random() * WORLD_HEIGHT;
      this.size = size || Math.random() * 4 + 3;
      this.color = color || colors[Math.floor(Math.random() * colors.length)];
    }

    draw() {
      ctx.beginPath();
      ctx.arc(this.x - camera.x, this.y - camera.y, this.size, 0, Math.PI * 2);
      ctx.fillStyle = this.color;
      ctx.shadowBlur = 8;
      ctx.shadowColor = this.color;
      ctx.fill();
      ctx.closePath();
      ctx.shadowBlur = 0;
    }
  }

  function init() {
    isGameOver = false;
    isBoosting = false;
    playerKills = 0;
    killVal.textContent = playerKills;
    gameOverScreen.style.display = 'none';
    snakes = [];
    foods = [];
    keysPressed = {};

    // Criar Jogador
    player = new Snake(WORLD_WIDTH / 2, WORLD_HEIGHT / 2, '#00ffcc', 'Você', false);
    snakes.push(player);

    // Criar Bots
    const botNames = ['Venom', 'Viper', 'Anaconda', 'Python', 'Cobrinha99', 'SlitherKing', 'Slinky', 'Serrana', 'Cascavel', 'Coral', 'Naja', 'Sidewinder', 'Mamba', 'Basilisk', 'Kaa'];
    for (let i = 0; i < NUM_BOTS; i++) {
      snakes.push(
        new Snake(
          Math.random() * WORLD_WIDTH,
          Math.random() * WORLD_HEIGHT,
          colors[Math.floor(Math.random() * colors.length)],
          botNames[i % botNames.length]
        )
      );
    }

    // Espalhar comida
    for (let i = 0; i < NUM_FOOD; i++) {
      foods.push(new Food());
    }
  }

  // Captura do movimento e clique do mouse
  window.addEventListener('mousemove', (e) => {
    mouse.x = e.clientX;
    mouse.y = e.clientY;
  });

  window.addEventListener('mousedown', (e) => {
    if (e.button === 0) {
      isBoosting = true;
    }
  });

  window.addEventListener('mouseup', (e) => {
    if (e.button === 0) {
      isBoosting = false;
    }
  });

  // Captura do Teclado
  window.addEventListener('keydown', (e) => {
    keysPressed[e.code] = true;
    keysPressed[e.key] = true;
  });

  window.addEventListener('keyup', (e) => {
    keysPressed[e.code] = false;
    keysPressed[e.key] = false;
  });

  // Loop Principal
  function loop() {
    if (isGameOver) return;

    if (player) {
      let head = player.segments[0];
      camera.x = head.x - canvas.width / 2;
      camera.y = head.y - canvas.height / 2;
    }

    ctx.fillStyle = '#111';
    ctx.fillRect(0, 0, canvas.width, canvas.height);

    ctx.strokeStyle = '#222';
    ctx.lineWidth = 1;
    const gridSize = 100;
    const startX = Math.max(0, Math.floor(camera.x / gridSize) * gridSize);
    const startY = Math.max(0, Math.floor(camera.y / gridSize) * gridSize);
    const endX = Math.min(WORLD_WIDTH, camera.x + canvas.width);
    const endY = Math.min(WORLD_HEIGHT, camera.y + canvas.height);

    for (let x = startX; x < endX; x += gridSize) {
      ctx.beginPath();
      ctx.moveTo(x - camera.x, 0);
      ctx.lineTo(x - camera.x, canvas.height);
      ctx.stroke();
    }
    for (let y = startY; y < endY; y += gridSize) {
      ctx.beginPath();
      ctx.moveTo(0, y - camera.y);
      ctx.lineTo(canvas.width, y - camera.y);
      ctx.stroke();
    }

    ctx.strokeStyle = '#ff3366';
    ctx.lineWidth = 5;
    ctx.strokeRect(0 - camera.x, 0 - camera.y, WORLD_WIDTH, WORLD_HEIGHT);

    // Atualizar e Desenhar Cobras
    for (let i = snakes.length - 1; i >= 0; i--) {
      let snake = snakes[i];
      snake.update();
      snake.draw();

      let head = snake.segments[0];

      for (let j = 0; j < snakes.length; j++) {
        if (i === j) continue;
        let otherSnake = snakes[j];

        for (let k = 0; k < otherSnake.segments.length; k++) {
          if (k === 0 && Math.hypot(head.x - otherSnake.segments[0].x, head.y - otherSnake.segments[0].y) > 20) continue;

          let seg = otherSnake.segments[k];
          let dist = Math.hypot(head.x - seg.x, head.y - seg.y);

          if (dist < snake.radius + otherSnake.radius * 0.8) {
            // Detectar se a morte da outra cobra foi causada por uma colisão no corpo do jogador
            let killedByPlayer = (!otherSnake.isBot && snake.isBot) || (otherSnake.isBot && !snake.isBot && otherSnake === player);
            killSnake(snake, i, killedByPlayer);
            break;
          }
        }
      }
    }

    // Desenhar e gerenciar comidas
    for (let i = foods.length - 1; i >= 0; i--) {
      let f = foods[i];
      f.draw();

      for (let snake of snakes) {
        let head = snake.segments[0];
        let dist = Math.hypot(head.x - f.x, head.y - f.y);
        if (dist < snake.radius + f.size) {
          snake.length += f.size * 0.4;
          foods.splice(i, 1);
          foods.push(new Food());
          break;
        }
      }
    }

    if (player) {
      scoreVal.textContent = Math.floor(player.length);
    }
    updateLeaderboard();

    requestAnimationFrame(loop);
  }

  function killSnake(snake, index, killedByPlayer) {
    for (let s of snake.segments) {
      if (Math.random() < 0.4) {
        foods.push(new Food(s.x, s.y, 8, snake.color));
      }
    }

    // Incrementar pontuação se o jogador foi quem provocou a eliminação
    if (killedByPlayer && snake.isBot) {
      playerKills++;
      killVal.textContent = playerKills;
    }

    snakes.splice(index, 1);

    if (!snake.isBot) {
      isGameOver = true;
      finalScore.textContent = Math.floor(snake.length);
      finalKills.textContent = playerKills;
      gameOverScreen.style.display = 'block';
    } else {
      setTimeout(() => {
        if (!isGameOver) {
          snakes.push(
            new Snake(
              Math.random() * WORLD_WIDTH,
              Math.random() * WORLD_HEIGHT,
              colors[Math.floor(Math.random() * colors.length)],
              snake.name
            )
          );
        }
      }, 3000);
    }
  }

  function updateLeaderboard() {
    let sorted = [...snakes].sort((a, b) => b.length - a.length);
    leaderboardList.innerHTML = '';

    for (let i = 0; i < Math.min(5, sorted.length); i++) {
      let s = sorted[i];
      let li = document.createElement('li');
      li.style.color = s.color;
      li.innerHTML = `<strong>${s.name}</strong>: ${Math.floor(s.length)}`;
      leaderboardList.appendChild(li);
    }
  }

  playBtn.addEventListener('click', () => {
    init();
    loop();
  });

  init();
  loop();
</script>
</body>
</html>
"""
display(HTML(html_content))
```
